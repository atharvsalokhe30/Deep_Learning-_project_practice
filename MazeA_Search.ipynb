```python
def astar(maze, start, goal):

    rows = len(maze)
    cols = len(maze[0])

    open_list = []
    heapq.heappush(open_list, (0, start, [start], 0))

    visited = set()

    directions = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    while open_list:

        f, current, path, g = heapq.heappop(open_list)

        print("\nSelected Cell:", current)
        print("Selected Path:", path)
        print("Selected f =", f, "and g =", g)

        if current == goal:
            print("\nGoal Reached")
            return path

        if current not in visited:

            visited.add(current)

            for dr, dc in directions:

                new_row = current[0] + dr
                new_col = current[1] + dc

                if (
                    0 <= new_row < rows and
                    0 <= new_col < cols and
                    maze[new_row][new_col] == 0 and
                    (new_row, new_col) not in visited
                ):

                    new_g = g + 1
                    new_h = heuristic((new_row, new_col), goal)
                    new_f = new_g + new_h

                    new_path = path + [(new_row, new_col)]

                    heapq.heappush(
                        open_list,
                        (
                            new_f,
                            (new_row, new_col),
                            new_path,
                            new_g
                        )
                    )

                    print("\nCandidate Path:", new_path)
                    print("g =", new_g, "h =", new_h, "f =", new_f)

    return None
```

```python
path = astar(maze, start, goal)

print("\nFinal Shortest Path:")
print(path)

print("Number of Steps:", len(path) - 1)
```
