```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def norm_func(i):
  x = (i-i.min())/(i.min()-i.max())
```

```python
#sklearn preprocessing library is used and minmaxscaler function is imported
from sklearn.preprocessing import MinMaxScaler
# initialized minmaxscaler function
trans = MinMaxScaler()
#fit this function and transform on data set
data = pd.DataFrame(trans.fit_transform(univ.iloc[:,1:]))
data
```

```python
from scipy.cluster.hierarchy import linkage
import scipy.cluster.hierarchy as sch
z=linkage(data, method="complete",metric="euclidean")

plt.figure(figsize=(15, 5))
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Index')
plt.ylabel('Distance')

sch.dendrogram(z,leaf_rotation=0.,leaf_font_size=8)
plt.show()
```
