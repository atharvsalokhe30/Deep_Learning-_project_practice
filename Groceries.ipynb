```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder

# Load the groceries dataset
with open('/content/groceries.csv', 'r') as f:
    transactions = f.read().splitlines()

# Preprocess the data into a list of lists (each sublist is a transaction)
transactions = [t.split(',') for t in transactions]

# Initialize TransactionEncoder
te = TransactionEncoder()

# One-hot encode the transactions
te_ary = te.fit(transactions).transform(transactions)
df_onehot = pd.DataFrame(te_ary, columns=te.columns_)

print(f"Shape of one-hot encoded DataFrame: {df_onehot.shape}")
display(df_onehot.head())
```

```python
from mlxtend.frequent_patterns import apriori

# Apply Apriori algorithm
frequent_itemsets = apriori(df_onehot, min_support=0.01, use_colnames=True)

print(f"Found {len(frequent_itemsets)} frequent itemsets.")
display(frequent_itemsets.head())
```

```python
from mlxtend.frequent_patterns import association_rules

# Generate rules with a minimum confidence threshold
# Even if ranking by lift, a confidence threshold helps filter out noise
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.1)

# Rank the rules by 'lift' in descending order
rules_ranked = rules.sort_values(by='lift', ascending=False)

# Display the top 10 rules
print(f"Generated {len(rules)} rules.")
display(rules_ranked.head(10))
```
