```python
#Import the libraries
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Import .csv file and convert it to a DataFrame object
df = pd.read_csv("/content/Iris.csv");
df
```

```python
print(df.info())
```

```python
df1 = df.drop(['Species'],axis=1)
```

```python
array=df1.values
array
```

```python
stscaler = StandardScaler().fit(array)
X = stscaler.transform(array)
X
```

```python
dbscan = DBSCAN(eps=0.45, min_samples=5)
dbscan.fit(X)
```

```python
#Noisy samples are given the label -1.
dbscan.labels_
```

```python
cl=pd.DataFrame(dbscan.labels_,columns=['cluster'])
```

```python
cl
pd.set_option("display.max_rows", None)
```

```python
cl
```

```python
df1 = pd.concat([df1,cl],axis=1)
df1
```

```python
import matplotlib.pyplot as plt
plt.style.use('bmh')
```

```python
df1.plot(x="Petal.Length",y ="Petal.Width",c=dbscan.labels_ ,kind="scatter",s=50 ,cmap=plt.cm.coolwarm)
plt.title('Clusters using DBScan')
```

```python
dl = dbscan.labels_
```

```python
from sklearn.metrics import silhouette_score
silhouette_score(X, dl)
```

```python
from sklearn.metrics import silhouette_score
silhouette_score(X, dl)
```

```python
from sklearn.cluster import KMeans
clf = KMeans(n_clusters=3)
y_kmeans = clf.fit_predict(X)
```

```python
cl1=pd.DataFrame(y_kmeans,columns=['Kcluster'])
cl1
```

```python
cl1=pd.DataFrame(y_kmeans,columns=['Kcluster'])
cl1
```

```python
df2 = pd.concat([df1,cl1],axis=1)
df2
```

```python
df2.plot(x="Petal.Length",y ="Petal.Width",c=y_kmeans ,kind="scatter",s=50 ,cmap=plt.cm.copper_r)
plt.title('Clusters using KMeans')
```
