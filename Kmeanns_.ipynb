```python
def norm_func(i):
    x = (i-i.min()) / (i.max() - i.min())
    return (x)
```

```python
df_norm = norm_func(Univ.iloc[:,1:])
```

```python
df_norm.head(10)  # Top 10 rows
```

```python
from sklearn.cluster import KMeans
fig = plt.figure(figsize=(10, 8))
WCSS = []     # Empty List
for i in range(1, 11):   # Loop runs K-Means for cluster numbers 1 to 10
    clf = KMeans(n_clusters=i) # Calculates Within-Cluster Sum of Squares (WCSS) for each K
    clf.fit(df_norm)  # It trains (fits) the K-Means model on the normalized dataset to find cluster centers and assign data points to clusters.
    WCSS.append(clf.inertia_) # inertia is another name for WCSS
plt.plot(range(1, 11), WCSS)
plt.title('The Elbow Method')
plt.ylabel('WCSS')
plt.xlabel('Number of Clusters')
plt.show()
```

```python
clf = KMeans(n_clusters=5)
y_kmeans = clf.fit_predict(df_norm)
```

```python
y_kmeans  # Display Cluster Labels
```

```python
#clf.cluster_centers_
clf.labels_
```

```python
md=pd.Series(y_kmeans)  # converting numpy array into pandas series object
Univ['clust']=md # creating a  new column and assigning it to new column
Univ
```

```python
Univ.iloc[:,1:7].groupby(Univ.clust).mean()
```

```python
Univ.plot(x="Top10",y ="SAT",c=clf.labels_,kind="scatter",s=50 ,cmap=plt.cm.coolwarm)
plt.title('Clusters using KMeans')
```
